#函数
作用：提高代码复用性
定义格式：
     def 函数名():
          函数体
调用格式：
     函数名()
*return返回多个值时，以元组形式返回。
retur与print区别：-1）return处停止执行，print会继续执行；
                 -2）return是返回值，函数使用return是要显示结果需要调用函数时加上print();函数                       直接使用print则调用时直接输出结果；
形参&实参
定义格式：
def 函数名(形参a，形参b):
    函数体

调用格式：
函数名(实参1，实参2)

*注意：定义与调用同级，不能在函数定义里调用他自己！！！
                 

#参数
-1）必备参数
含义：传递和定义参数的顺序与个数必须一致。
-2）默认参数
含义：调用时可以不传参数，使用默认参数。
*注意：调用时所有的位置参数放在默认参数之前。
-3）可变参数
含义：传入的值的数量可以改变，也可以不传，以元组形式接收。
格式：def 函数名(*形参):
-4)关键参数
含义：也是可变参数的一种。
格式：def 函数名(**形参):
*注意：传入值时以“键=值”的形式传入。


#作用域

-1）全局变量
函数外部定义的变量。
*函数体使用变量会先在内部找变量，有就直接使用，不会影响函数外对应的的变量。

-2）局部变量
函数内部对应的变量，从定义位置到函数定义结束有效。
作用：临时保存数据

-3）global
作用：函数内部声明全局变量
格式：global 变量名
      *注意：不能写为：global 变量名 = 值，要分开写。

-4）nolocal
作用：有函数嵌套时，内层函数变量用nolocal声明，可以修改外层函数对应的变量的值。

#匿名函数

基本格式：
函数名 = lambda 形参 : 返回值（表达式）
*注意：形参不用打括号了！！！

调用：
结果 = 函数名（实参）

应用：
可以减少代码量，但降低了代码的可读性。

```python
comp = lambda a,b : "a<b" if a<b else "a>b"
print(comp(8,5))
```

#拆包

含义：对于函数的多个返回值，去掉元组，列表，字典，直接得到里面的数据的过程。



```python
tua = [1,2,3,4]
#方法一
a,b,c,d = tua
print("a=",a,"b=",b,"c=",c,"d=",d)
#要求变量数量与元组元素个数一致，一般用在获取元组元素时

#方法二
a,*b = tua
c,*d = b
e,f = d
print(a,c,e,f)
#应用可变变量一层层剥离
```

#模块

就是.py文件，负责定义函数和变量，使用时导入模块，方便维护，可重复使用，也避免函数名与变量名冲突。

分类：
1.内置模块（标准库模块）
2.第三方模块：通过包管理工具pip安装
3.自定义模块：模块名要遵循python的命名规则。

使用import导入模块：
1.导入格式：
import 模块名
2.调用格式：
模块名.功能名；模块名.变量名。

从模块导入指定的部分：
from 模块名 improt 功能1，功能2，...
*调用时直接使用功能名，不用加模块名。

as起别名：
import 模块名 as 别名
from 模块名 import 功能名1 as 别名，功能名2 as 别名

内置全局变量__name__
语法格式：
if__name__=="__main__"：
     内容
     
作用：
控制py文件在不同场景应用不同逻辑。
*模块导入时，if__name__=="__main__"：下的内容不会显示调用

#包（Package）

含义：项目结构的文件夹目录。
*包自带一个__init__.py的文件夹。

作用：将有联系的模块放到同一个文件夹，避免模块名起冲突，让结构更清晰。

新建包：右键项目名->new->python package

*import导入包时，会先执行__init__.py的文件代码。（不建议在这个文件中写大量代码）

使用方式：在init文件中写导入包的其它模块的代码，在主文件要用时就只用导入包：import pack_01

__all__:
本质上是一个列表，可以控制要导入的模块：
__all__ = [(要导入的模块)]
在init文件写这个代码可以控制主文件导入包时要导入的模块。

*包本质是一个模块，包可以又包含包。

#闭包

作用：保存函数内的变量在使用完后不被销毁

条件：1.嵌套函数定义；2.内层函数使用外层函数的变量；3.外层函数的返回值是内层函数的函数名



```python
def outer(m):
    print("外层函数的值：",m)
    def inner(n):
        print("内层函数的值：",n)
        return m+n
    return inner

ot = outer(10)

#第一次调用内层函数
print(ot(20))
#第二次调用内层函数
print(ot(40))

#实现了外层函数的闭包变量的保存，每次使用的都是同一个闭包变量！！！
```
