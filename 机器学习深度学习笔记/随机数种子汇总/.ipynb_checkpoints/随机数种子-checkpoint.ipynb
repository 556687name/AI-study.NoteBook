### 为什么参数名叫 `random_state` 而不是 `seed`？

只是**命名习惯不同**，意思完全一样：

| 库 | 参数名 | 本质 |
|---|---|---|
| scikit-learn | `random_state` | 随机种子 |
| NumPy | `random.seed()` | 随机种子 |
| PyTorch | `manual_seed()` | 随机种子 |

scikit-learn 选择叫 `random_state` 是为了表达更清晰：它设置的是**随机数生成器的状态**。

### `42` 这个数字有什么特殊？

这是一个经典的"梗"——出自科幻小说《银河系漫游指南》，书中说 **42** 是"生命、宇宙以及一切的终极答案"。

程序员们喜欢用它作为随机种子，纯粹是玩梗 + 约定俗成，用 `0`、`1`、`123` 也一样。
