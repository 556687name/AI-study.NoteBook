# Python 线性代数常见写法（NumPy & SciPy）

---

## 一、向量与矩阵基础

```python
import numpy as np

# 向量
v = np.array([1, 2, 3])              # 行向量（一维）
u = np.array([[1], [2], [3]])        # 列向量（二维）

# 零向量 / 单位向量
np.zeros(3)                           # [0. 0. 0.]
np.ones(3)                            # [1. 1. 1.]

# 矩阵
A = np.array([[1, 2], [3, 4]])
B = np.zeros((3, 3))                  # 零矩阵
C = np.ones((2, 4))                   # 全1矩阵
I = np.eye(3)                         # 3×3 单位矩阵
D = np.diag([1, 2, 3])               # 对角矩阵

# 随机矩阵
R = np.random.rand(3, 3)              # [0,1) 均匀分布
Rn = np.random.randn(3, 3)            # 标准正态分布
```

---

## 二、矩阵运算

```python
A @ B           # 矩阵乘法（推荐写法，Python 3.5+）
np.dot(A, B)    # 矩阵乘法 / 向量内积
np.matmul(A, B) # 矩阵乘法（仅矩阵，不广播）

A * B           # 逐元素乘法（Hadamard积）
A ** 2          # 逐元素平方

A.T             # 转置
A.conj().T      # 共轭转置（Hermitian）
np.transpose(A) # 转置

np.trace(A)     # 迹（对角线元素之和）
np.sum(np.diag(A))  # 等价于 trace
```

---

## 三、行列式、秩、逆

```python
np.linalg.det(A)        # 行列式 |A|
np.linalg.matrix_rank(A) # 秩
np.linalg.inv(A)         # 逆矩阵 A⁻¹
np.linalg.pinv(A)        # 伪逆（Moore-Penrose），支持非方阵
```

---

## 四、特征值与特征向量

```python
# 特征值与特征向量
eigenvalues, eigenvectors = np.linalg.eig(A)
# A @ v = λ @ v, eigenvectors 每列是一个特征向量

# 仅特征值
np.linalg.eigvals(A)

# 对称/Hermitian 矩阵（更快更稳定）
eigenvalues, eigenvectors = np.linalg.eigh(A)
```

---

## 五、范数

```python
np.linalg.norm(v)        # L2 范数（默认）
np.linalg.norm(v, ord=1)  # L1 范数
np.linalg.norm(v, ord=np.inf)  # L∞ 范数
np.linalg.norm(A, ord='fro')   # Frobenius 范数（矩阵）
np.linalg.norm(A, ord=2)       # 谱范数（最大奇异值）
```

---

## 六、分解（Factorizations）

```python
# LU 分解 → PA = LU
import scipy.linalg as la
P, L, U = la.lu(A)

# QR 分解 → A = QR
Q, R = np.linalg.qr(A)

# Cholesky 分解（正定矩阵）→ A = L @ L.T
L = np.linalg.cholesky(A)

# SVD 分解 → A = U @ S @ Vh
U, S, Vh = np.linalg.svd(A)

# 从奇异值重建对角矩阵
Sigma = np.zeros(A.shape)
np.fill_diagonal(Sigma, S)
```

---

## 七、解线性方程组

```python
# 求解 Ax = b（推荐，比 inv 更稳定）
x = np.linalg.solve(A, b)

# 最小二乘解（超定 / 欠定）
x, residuals, rank, s = np.linalg.lstsq(A, b)

# 求 A 的逆再乘（不推荐，数值不稳定）
x = np.linalg.inv(A) @ b
```

## 九、内积 / 外积

```python
# 向量内积（点积）
np.dot(u, v)
u @ v
np.inner(u, v)

# 向量外积
np.outer(u, v)

# 两个向量夹角余弦
cos = np.dot(u, v) / (np.linalg.norm(u) * np.linalg.norm(v))
```

## 十、条件数

```python
# 条件数（衡量数值稳定性，越大越病态）
np.linalg.cond(A)       # 默认 L2 条件数
np.linalg.cond(A, 1)    # L1 条件数
np.linalg.cond(A, np.inf)  # L∞ 条件数
```

# 矩阵求导公式大全

矩阵求导是机器学习和优化中的核心数学工具。下面系统梳理常见公式。

---

## 一、基本约定

| 符号 | 含义 |
|------|------|
| $\mathbf{x}$ | 列向量 $\in \mathbb{R}^n$ |
| $\mathbf{A}$ | 矩阵 $\in \mathbb{R}^{m \times n}$（常数） |
| $\mathbf{a}$ | 常数向量 |
| $f(\mathbf{x})$ | 标量函数 |

> **布局约定**：以下采用**分母布局（Denominator Layout）**，即 $\frac{\partial f}{\partial \mathbf{x}}$ 结果为**列向量**，与 $\mathbf{x}$ 同形状。

---

## 二、标量对向量求导

### 1. 线性形式

$$\frac{\partial (\mathbf{a}^T \mathbf{x})}{\partial \mathbf{x}} = \frac{\partial (\mathbf{x}^T \mathbf{a})}{\partial \mathbf{x}} = \mathbf{a}$$

```python
# 例：f = 3x₁ + 4x₂ → 梯度 = [3, 4]ᵀ
```

### 2. 二次型

$$\frac{\partial (\mathbf{x}^T \mathbf{A} \mathbf{x})}{\partial \mathbf{x}} = (\mathbf{A} + \mathbf{A}^T)\mathbf{x}$$

若 $\mathbf{A}$ 对称：$= 2\mathbf{A}\mathbf{x}$

```python
# 例：f = xᵀAx, A对称 → 梯度 = 2Ax
```

### 3. 常见组合

| 函数 | 梯度 |
|------|------|
| $\mathbf{b}^T \mathbf{x}$ | $\mathbf{b}$ |
| $\mathbf{x}^T \mathbf{x}$ (即 $\|\mathbf{x}\|_2^2$) | $2\mathbf{x}$ |
| $\|\mathbf{A}\mathbf{x} - \mathbf{b}\|_2^2$ | $2\mathbf{A}^T(\mathbf{A}\mathbf{x} - \mathbf{b})$ |
| $\mathbf{x}^T \mathbf{A} \mathbf{x}$ | $(\mathbf{A} + \mathbf{A}^T)\mathbf{x}$ |

---

## 三、向量对向量求导（Jacobian）

$$\frac{\partial \mathbf{y}}{\partial \mathbf{x}} = \begin{bmatrix} 
\frac{\partial y_1}{\partial x_1} & \cdots & \frac{\partial y_1}{\partial x_n} \\
\vdots & \ddots & \vdots \\
\frac{\partial y_m}{\partial x_1} & \cdots & \frac{\partial y_m}{\partial x_n}
\end{bmatrix}_{m \times n}$$

### 常用

$$\frac{\partial (\mathbf{A}\mathbf{x})}{\partial \mathbf{x}} = \mathbf{A}$$

$$\frac{\partial (\mathbf{x}^T \mathbf{A})}{\partial \mathbf{x}} = \mathbf{A}^T$$

$$\frac{\partial (\mathbf{A}\mathbf{x} + \mathbf{b})}{\partial \mathbf{x}} = \mathbf{A}$$

---

## 四、标量对矩阵求导

$$\frac{\partial f}{\partial \mathbf{X}} = \begin{bmatrix}
\frac{\partial f}{\partial X_{11}} & \cdots & \frac{\partial f}{\partial X_{1n}} \\
\vdots & \ddots & \vdots \\
\frac{\partial f}{\partial X_{m1}} & \cdots & \frac{\partial f}{\partial X_{mn}}
\end{bmatrix}$$

### 常用公式

| $f(\mathbf{X})$ | $\frac{\partial f}{\partial \mathbf{X}}$ |
|-----------------|------------------------------------------|
| $\mathbf{a}^T \mathbf{X} \mathbf{b}$ | $\mathbf{a}\mathbf{b}^T$ |
| $\mathbf{a}^T \mathbf{X}^T \mathbf{b}$ | $\mathbf{b}\mathbf{a}^T$ |
| $\text{tr}(\mathbf{X})$ | $\mathbf{I}$ |
| $\text{tr}(\mathbf{A}\mathbf{X})$ | $\mathbf{A}^T$ |
| $\text{tr}(\mathbf{X}^T \mathbf{A})$ | $\mathbf{A}$ |
| $\text{tr}(\mathbf{X}^T \mathbf{X})$ | $2\mathbf{X}$ |
| $\text{tr}(\mathbf{X}^T \mathbf{A} \mathbf{X})$ | $(\mathbf{A} + \mathbf{A}^T)\mathbf{X}$ |
| $\text{tr}(\mathbf{X}^{-1}\mathbf{A})$ | $-\mathbf{X}^{-T}\mathbf{A}^T\mathbf{X}^{-T}$ |
| $\|\mathbf{X}\|_F^2$ | $2\mathbf{X}$ |
| $\det(\mathbf{X})$ | $\det(\mathbf{X}) \cdot \mathbf{X}^{-T}$ |
| $\ln\det(\mathbf{X})$ | $\mathbf{X}^{-T}$ |

---

## 五、激活函数相关（ML常用）

| 函数 | 导数 |
|------|------|
| $\sigma(x) = \frac{1}{1+e^{-x}}$ | $\sigma(x)(1-\sigma(x))$ |
| $\tanh(x)$ | $1 - \tanh^2(x)$ |
| $\text{ReLU}(x)$ | $\begin{cases} 1 & x>0 \\ 0 & x \le 0 \end{cases}$ |
| $\text{softmax}(\mathbf{x})_i$ | $\text{softmax}_i(\delta_{ij} - \text{softmax}_j)$ |

---

## 六、损失函数梯度

### MSE（均方误差）

$$L = \frac{1}{n}\|\mathbf{X}\mathbf{w} - \mathbf{y}\|_2^2$$

$$\frac{\partial L}{\partial \mathbf{w}} = \frac{2}{n}\mathbf{X}^T(\mathbf{X}\mathbf{w} - \mathbf{y})$$

### 交叉熵 + Softmax

$$\frac{\partial L}{\partial \mathbf{z}} = \hat{\mathbf{y}} - \mathbf{y}$$

（$\mathbf{z}$ 为 logits，$\hat{\mathbf{y}}$ 为 softmax 输出，$\mathbf{y}$ 为 one-hot 标签）

---

## 七、实用法则

### 链式法则

$$\frac{\partial f(g(\mathbf{x}))}{\partial \mathbf{x}} = \frac{\partial g}{\partial \mathbf{x}} \cdot \frac{\partial f}{\partial g}$$

### 乘积法则

$$\frac{\partial (\mathbf{u}^T \mathbf{v})}{\partial \mathbf{x}} = \mathbf{u}^T \frac{\partial \mathbf{v}}{\partial \mathbf{x}} + \mathbf{v}^T \frac{\partial \mathbf{u}}{\partial \mathbf{x}}$$
