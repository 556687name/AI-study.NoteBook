# BGD批量梯度下降

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
X = np.random.rand(100,1)

w,b = np.random.randint(1,10,size=2)

y = w*X+b+np.random.randn(100,1)
plt.scatter(X,y)
```

```python
X_b=np.concatenate([X,np.full(shape =(100,1),fill_value=1)],axis=1)
X[:10]
```

```python
epoches = 10000

eta = 0.01
t0 = 5
t1 = 1000
def learning_rate_shedule(t):
    return t0/(t+t1)

theta = np.random.randn(2,1)

t = 0
for i in range(epoches):
    g = X_b.T.dot(X_b.dot(theta)-y)
    eta = learning_rate_shedule(t)
    theta = theta - eta*g
    t += 1

print("真实的w,b:",w,b)
print("预测的w,b:",theta)
```

# SGD随机梯度下降

```python
import numpy as np
import matplotlib.pyplot as plt

X = np.random.rand(100,1)
w,b = np.random.randint(1,10,size=2)
y = w*X+b+np.random.randn(100,1)
plt.scatter(X,y)
```

```python
epoches = 100

eta = 0.01
t0 = 5
t1 = 1000
def learning_rate_shedule(t):
    return t0/(t+t1)

theta = np.random.randn(2,1)

count = 0
for t in range(epoches):
    index = np.arange(100)
    np.random.shuffle(index)

    X =X[index]
    y =y[index]
    for i in range(100):
        X_i = X[[i]]
        X_i = np.concatenate([X_i, np.full(shape=(1, 1), fill_value=1)], axis=1)
        y_i = y[[i]]

        g = X_i.T.dot(X_i.dot(theta) - y_i)
        eta = learning_rate_shedule(count)
        count += 1
        theta -=eta*g

print("真实的w,b:",w,b)
print("预测的w,b:",theta)
```

# MBGD小批量梯度下降

```python
import numpy as np
import matplotlib.pyplot as plt
X = np.random.rand(100,1)
w,b = np.random.randint(1,10,size=2)
y = w*X+b+np.random.randn(100,1)
plt.scatter(X,y)

X_b=np.concatenate([X,np.full(shape =(100,1),fill_value=1)],axis=1)

t0 = 5
t1 = 1000
def learning_rate_shedule(t):
    return t0/(t+t1)
```

```python
epoches = 100
n = 100
batch_size = 16
num_batches = int(n/batch_size)
```

```python
theta = np.random.randn(2,1)
```

```python
for epoch in range(epoches):
    index = np.arange(n)
    np.random.shuffle(index)
    X_b =X_b[index]
    y =y[index]
    for i in range(num_batches):
        X_b_batch = X_b[i*batch_size:(i+1)*batch_size]
        y_batch = y[i*batch_size:(i+1)*batch_size]
        g = X_b_batch.T.dot(X_b_batch.dot(theta)-y_batch)
        learning_rate = learning_rate_shedule(epoch*n+i)
        theta = theta - learning_rate * g

print("真实的w,b:",w,b)
print("预测的w,b:",theta)
```
