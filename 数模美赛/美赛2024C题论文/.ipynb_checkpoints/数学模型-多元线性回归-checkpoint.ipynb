# 多元线性回归

## 线性模型
<font size="3">
    1)给定n维输入 <b>$$X=[x_1,x_2,...,x_n]^T$$
    </b></font>
<font size="3">
    2)线性模型有一个n维权重和一个标量偏差<b>$$w=[w_1,w_2,...,w_n]^T,b$$
    </b></font>
<font size="3">
    3)输出是输入的加权<b>$$y=w_1x_1+w_2x_2+...+w_nx_n+b$$
    </b></font>
<div style="font-size:20px; color:blue;">
    向量版本：y = $\langle w, x \rangle$+b
</div>

<div style="font-size:15px; color:gray;">
【注】线性回归模型是一个单层神经网络
</div>

## 衡量预估质量
<font size="3">1.比较真实值和估计值
</font>
<font size="3">2.假设y是真实值，$\hat{y}$是估计值，我们可以比较：<b>$$\text{MSE} = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2$$
</b>这个叫做平方损失（误差）</font>

## 常用误差函数速查表

在机器学习中，误差函数（损失函数）用来衡量模型预测值 \(\hat{y}\) 与真实值 \(y\) 之间的差距，模型训练的目标就是最小化这个误差。

| 名称 | 含义 | LaTeX 公式 | 用途 |
|------|------|-----------|------|
| MSE（均方误差） | 预测值与真实值之差的平方的均值，对大误差惩罚更重 | `\frac{1}{n}\sum_{i=1}^{n}(y_i - \hat{y}_i)^2` | 回归 |
| RMSE（均方根误差） | MSE 的平方根，量纲与原始数据一致 | `\sqrt{\frac{1}{n}\sum_{i=1}^{n}(y_i - \hat{y}_i)^2}` | 回归 |
| MAE（平均绝对误差） | 预测值与真实值之差的绝对值的均值，对异常值更鲁棒 | `\frac{1}{n}\sum_{i=1}^{n}\|y_i - \hat{y}_i\|` | 回归 |
| SSE（残差平方和） | 所有样本预测误差的平方之和 | `\sum_{i=1}^{n}(y_i - \hat{y}_i)^2` | 回归 |
| 交叉熵损失 | 衡量两个概率分布之间的差异，常用于分类任务 | `-\frac{1}{n}\sum_{i=1}^{n}[y_i\log(\hat{y}_i)+(1-y_i)\log(1-\hat{y}_i)]` | 分类 |

## 常用误差函数速查表



| 名称 | 含义 | 公式 | 用途 |
|------|------|------|------|
| MSE（均方误差） | 对大误差惩罚更重 | $\text{MSE} = \frac{1}{n}\sum_{i=1}^{n}(y_i - \hat{y}_i)^2$ | 回归 |
| RMSE（均方根误差） | 量纲与原始数据一致 | $\text{RMSE} = \sqrt{\frac{1}{n}\sum_{i=1}^{n}(y_i - \hat{y}_i)^2}$ | 回归 |
| MAE（平均绝对误差） | 对异常值更鲁棒 | $\text{MAE} = \frac{1}{n}\sum_{i=1}^{n}\|y_i - \hat{y}_i\|$ | 回归 |
| SSE（残差平方和） | 所有误差平方之和 | $\text{SSE} = \sum_{i=1}^{n}(y_i - \hat{y}_i)^2$ | 回归 |
| 交叉熵损失 | 衡量概率分布差异 | $\text{CE} = -\frac{1}{n}\sum_{i=1}^{n}[y_i\log(\hat{y}_i)+(1-y_i)\log(1-\hat{y}_i)]$ | 分类 |

## 参数学习
<font size="3"> 
1.训练损失<b>$$\text{MSE} = \frac{1}{2n} \sum_{i=1}^{n} (y_i - \langle w, x_i \rangle$-b)^2=\frac{1}{2n}||y-Xw-b||^2$$</b></font>
    <font size="3">2.最小化损失来学习参数$$w*,b* = argmin_{w,b}\text{MSE}$$
</b></font>


## 显示解 (Closed-form Solution)

- **将偏差加入权重**
  $$ \mathbf{X} \leftarrow [\mathbf{X}, \mathbf{1}], \quad \mathbf{w} \leftarrow \begin{bmatrix} \mathbf{w} \\ b \end{bmatrix} $$

  **损失函数与梯度：**
  $$ \ell(\mathbf{X}, \mathbf{y}, \mathbf{w}) = \frac{1}{2n} \| \mathbf{y} - \mathbf{X}\mathbf{w} \|^2 $$
  $$ \frac{\partial}{\partial \mathbf{w}} \ell(\mathbf{X}, \mathbf{y}, \mathbf{w}) = \frac{1}{n} (\mathbf{y} - \mathbf{X}\mathbf{w})^T \mathbf{X} $$

- **损失是凸函数，所以最优解满足**

  令梯度为 0：
  $$ \frac{\partial}{\partial \mathbf{w}} \ell(\mathbf{X}, \mathbf{y}, \mathbf{w}) = 0 $$

  推导过程：
  $$ \Leftrightarrow \frac{1}{n} (\mathbf{y} - \mathbf{X}\mathbf{w})^T \mathbf{X} = 0 $$

  最终解（正规方程）：
  $$ \Leftrightarrow \mathbf{w}^* = (\mathbf{X}^T\mathbf{X})^{-1} \mathbf{X}^T\mathbf{y} $$
