# 多元线性回归

## 线性模型
<font size="3">
    1)给定n维输入 <b>$$X=[x_1,x_2,...,x_n]^T$$
    </b></font>
<font size="3">
    2)线性模型有一个n维权重和一个标量偏差<b>$$w=[w_1,w_2,...,w_n]^T,b$$
    </b></font>
<font size="3">
    3)输出是输入的加权<b>$$y=w_1x_1+w_2x_2+...+w_nx_n+b$$
    </b></font>
<div style="font-size:20px; color:blue;">
    向量版本：y = $\langle w, x \rangle$+b
</div>

<div style="font-size:15px; color:gray;">
【注】线性回归模型是一个单层神经网络
</div>

## 衡量预估质量
<font size="3">1.比较真实值和估计值
</font>
<font size="3">2.假设y是真实值，$\hat{y}$是估计值，我们可以比较：<b>$$\text{MSE} = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2$$
</b>这个叫做平方损失（误差）</font>

## 常用误差函数速查表

在机器学习中，误差函数（损失函数）用来衡量模型预测值 \(\hat{y}\) 与真实值 \(y\) 之间的差距，模型训练的目标就是最小化这个误差。

| 名称 | 含义 | LaTeX 公式 | 用途 |
|------|------|-----------|------|
| MSE（均方误差） | 预测值与真实值之差的平方的均值，对大误差惩罚更重 | `\frac{1}{n}\sum_{i=1}^{n}(y_i - \hat{y}_i)^2` | 回归 |
| RMSE（均方根误差） | MSE 的平方根，量纲与原始数据一致 | `\sqrt{\frac{1}{n}\sum_{i=1}^{n}(y_i - \hat{y}_i)^2}` | 回归 |
| MAE（平均绝对误差） | 预测值与真实值之差的绝对值的均值，对异常值更鲁棒 | `\frac{1}{n}\sum_{i=1}^{n}\|y_i - \hat{y}_i\|` | 回归 |
| SSE（残差平方和） | 所有样本预测误差的平方之和 | `\sum_{i=1}^{n}(y_i - \hat{y}_i)^2` | 回归 |
| 交叉熵损失 | 衡量两个概率分布之间的差异，常用于分类任务 | `-\frac{1}{n}\sum_{i=1}^{n}[y_i\log(\hat{y}_i)+(1-y_i)\log(1-\hat{y}_i)]` | 分类 |

## 常用误差函数速查表



| 名称 | 含义 | 公式 | 用途 |
|------|------|------|------|
| MSE（均方误差） | 对大误差惩罚更重 | $\text{MSE} = \frac{1}{n}\sum_{i=1}^{n}(y_i - \hat{y}_i)^2$ | 回归 |
| RMSE（均方根误差） | 量纲与原始数据一致 | $\text{RMSE} = \sqrt{\frac{1}{n}\sum_{i=1}^{n}(y_i - \hat{y}_i)^2}$ | 回归 |
| MAE（平均绝对误差） | 对异常值更鲁棒 | $\text{MAE} = \frac{1}{n}\sum_{i=1}^{n}\|y_i - \hat{y}_i\|$ | 回归 |
| SSE（残差平方和） | 所有误差平方之和 | $\text{SSE} = \sum_{i=1}^{n}(y_i - \hat{y}_i)^2$ | 回归 |
| 交叉熵损失 | 衡量概率分布差异 | $\text{CE} = -\frac{1}{n}\sum_{i=1}^{n}[y_i\log(\hat{y}_i)+(1-y_i)\log(1-\hat{y}_i)]$ | 分类 |

## 参数学习
<font size="3"> 
1.训练损失<b>$$\text{MSE} = \frac{1}{2n} \sum_{i=1}^{n} (y_i - \langle w, x_i \rangle$-b)^2=\frac{1}{2n}||y-Xw-b||^2$$</b></font>
    <font size="3">2.最小化损失来学习参数$$w*,b* = argmin_{w,b}\text{MSE}$$
</b></font>


## 显示解 (Closed-form Solution)

- **将偏差加入权重**
  $$ \mathbf{X} \leftarrow [\mathbf{X}, \mathbf{1}], \quad \mathbf{w} \leftarrow \begin{bmatrix} \mathbf{w} \\ b \end{bmatrix} $$

  **损失函数与梯度：**
  $$ \ell(\mathbf{X}, \mathbf{y}, \mathbf{w}) = \frac{1}{2n} \| \mathbf{y} - \mathbf{X}\mathbf{w} \|^2 $$
  $$ \frac{\partial}{\partial \mathbf{w}} \ell(\mathbf{X}, \mathbf{y}, \mathbf{w}) = \frac{1}{n} (\mathbf{y} - \mathbf{X}\mathbf{w})^T \mathbf{X} $$

- **损失是凸函数，所以最优解满足**

  令梯度为 0：
  $$ \frac{\partial}{\partial \mathbf{w}} \ell(\mathbf{X}, \mathbf{y}, \mathbf{w}) = 0 $$

  推导过程：
  $$ \Leftrightarrow \frac{1}{n} (\mathbf{y} - \mathbf{X}\mathbf{w})^T \mathbf{X} = 0 $$

  最终解（正规方程）：
  $$ \Leftrightarrow \mathbf{w}^* = (\mathbf{X}^T\mathbf{X})^{-1} \mathbf{X}^T\mathbf{y} $$

## 一、导入库 & 读数据

```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
import matplotlib.pyplot as plt
import seaborn as sns

# 设置中文显示
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False

df = pd.read_csv('data.csv')
```

---

## 二、数据预处理

### 1. 看概况 & 缺失值

```python
print(df.info())
print(df.describe())
print(df.isnull().sum())
```

### 2. 处理缺失值

```python
df = df.dropna()                    # 简单粗暴：删掉
# df['col'].fillna(df['col'].median(), inplace=True)  # 或用中位数填充
```

### 3. 异常值处理（画箱线图看）

```python
sns.boxplot(data=df)
plt.show()
# 3σ原则或IQR法剔除极端值
```

### 4. 分类变量转哑变量 一句话：把"红黄蓝"变成"是不是红？是不是黄？"两列 0/1，蓝就是两个 0。

```python
df = pd.get_dummies(df, columns=['类别列名'], drop_first=True)
```

---

## 三、相关性分析（选变量前必做）

```python
# 热力图
corr_matrix = df.corr()
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('相关性热力图')
plt.show()

# 看各变量与 y 的相关系数
print(corr_matrix['y'].sort_values(ascending=False))
```

---

## 四、设定 X 和 y

```python
y = df['y']
X = df.drop(columns=['y'])

# 【关键】添加截距项（常数列）
X = sm.add_constant(X)
```

> ⚠️ 不加 `add_constant` 会强制过原点，R² 都可能算错。

---

## 五、拟合模型（OLS）

```python
model = sm.OLS(y, X)
result = model.fit()
print(result.summary())
```

---

## 六、读 summary 的关键指标

```
                            OLS Regression Results
==============================================================================
Dep. Variable:                      y   R-squared:                       0.821
Model:                            OLS   Adj. R-squared:                  0.818
Method:                 Least Squares   F-statistic:                     412.3
Prob (F-statistic):           1.23e-56   Log-Likelihood:                -142.56
No. Observations:                 200   AIC:                             293.1
Df Residuals:                     196   BIC:                             306.3
==============================================================================
                 coef    std err          t      P>|t|      [0.025      0.975]
------------------------------------------------------------------------------
const          2.9821      0.104     28.695      0.000       2.777       3.187
x1             1.5123      0.051     29.653      0.000       1.412       1.613
x2            -0.7890      0.035    -22.543      0.000      -0.858      -0.720
==============================================================================
```

### 必看项：

| 指标 | 位置 | 判断标准 |
|:--|:--|:--|
| **调整 R²** | 右上 | 越高越好，> 0.3 模型有意义 |
| **F 统计量 p 值** | Prob (F) | < 0.05 模型整体显著 |
| **各系数 p 值** | P>\|t\| 列 | < 0.05 该变量显著 |
| **系数符号** | coef | 看正负是否和实际意义一致 |

---

## 七、逐步回归 / 变量筛选（p 值不显著时用）

### 方法一：手动剔除

```python
# 看 p 值最大的变量，删掉，重新拟合
X_new = X.drop(columns=['不显著的变量'])
model2 = sm.OLS(y, X_new).fit()
print(model2.summary())
```

### 方法二：用 sklearn 的 RFE 自动选

```python
from sklearn.feature_selection import RFE
from sklearn.linear_model import LinearRegression

selector = RFE(LinearRegression(), n_features_to_select=5)
selector.fit(X, y)
selected = X.columns[selector.support_]
print('保留的变量：', selected.tolist())
```

> RFE 只是帮你筛变量，最终还是要回 statsmodels 做显著性检验。

---

## 八、多重共线性检验（VIF）

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor

vif_data = pd.DataFrame({
    '变量': X.columns,
    'VIF': [variance_inflation_factor(X.values, i) for i in range(X.shape[1])]
})
print(vif_data)
```

| VIF | 含义 |
|:--|:--|
| < 5 | 没问题 |
| 5~10 | 轻度共线性 |
| > 10 | 严重共线性，该变量要删掉或合并 |

---

## 九、残差诊断（证明模型能用）

### 1. 正态性（残差服从正态分布）

```python
residuals = result.resid
sm.qqplot(residuals, line='45')
plt.title('残差 Q-Q 图')
plt.show()
```

点在 45° 线附近 → 正态性满足。

### 2. 异方差检验（残差方差恒定）

```python
# Breusch-Pagan 检验
from statsmodels.stats.diagnostic import het_breuschpagan

bp_test = het_breuschpagan(residuals, result.model.exog)
print(f'BP 检验 p 值：{bp_test[1]:.4f}')
```

| p 值 | 含义 |
|:--|:--|
| > 0.05 | 无异方差，好 ✓ |
| < 0.05 | 有异方差，需要用稳健标准误 |

### 3. 有异方差怎么办

```python
# 用 HC 稳健标准误重新估计
result_robust = model.fit(cov_type='HC3')
print(result_robust.summary())
```

### 4. 自相关检验（时间序列才需要）

```python
# Durbin-Watson 统计量在 summary 左上角就有
# DW ≈ 2  → 无自相关 ✓
# DW < 1 或 > 3 → 有自相关 ✗
```

```
               y         x1         x2         x3
count  200.0000  200.0000  200.0000  195.0000   ← x3 只剩 195
mean    15.2034    8.7712    3.1298   52.3400
std      5.8213    2.3341    1.2021   18.7723
min      2.1000    3.2000    0.5000   18.0000
25%     11.4500    7.1000    2.3000   38.5000
50%     15.1000    8.8000    3.1000   52.0000
75%     19.2000   10.4000    3.9000   66.0000
max     29.8000   15.1000    6.2000   98.0000   ← 有没有离谱值？
```

### 3. `df.isnull().sum()` — 看缺失

```python
print(df.isnull().sum())
```

输出示例：

```
y      0
x1     0
x2     0
x3     5
x4     0
dtype: int64
```

> 一目了然：x3 缺了 5 个值，别的都没问题。

---

### 三行联动逻辑

```
info()    → 发现 x3 有缺失、发现 类别 是 object
describe()→ 发现 x3 的 max 可能异常
isnull()  → 确认 x3 缺了 5 行
```

然后你就能决定：

```python
df = df.dropna()                              # 删缺失行
# 或
df['x3'] = df['x3'].fillna(df['x3'].median()) # 中位数填充
```

---

> 一句话总结：**这三行就像体检——先看有多少器官（info），再看器官指标（describe），最后查哪里有问题（isnull）。**
