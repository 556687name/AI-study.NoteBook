```python
class GameCharacter:
    def __init__(self,name,hp):
        self.name = name
        self.hp = hp

    def __str__(self):
        return(f"[角色]{self.name}|剩余HP:{self.hp}")
    
    
    def take_damage(self,damage):
        self.hp-=damage
        print("%s受到了%d点伤害，剩余HP：%d"%(self.name,damage,self.hp))

    def attack(self,target):
        print(f"{self.name}对{target.name}发动了普通攻击！")
        target.take_damage(10)  # 让目标承受 10 点伤害

    def  is_alive(self):
        return self.hp > 0


class Warrior(GameCharacter):
    def __init__(self,name):
        super().__init__(name,150)
        self.armor = 10

    def __str__(self):
        return f"[战士] {self.name} | HP: {self.hp} | 护甲: {self.armor}"

    
        
    def attack(self,target):
        print(f"{self.name}发动了飞龙斩！")
        target.take_damage(20)

    

class Mage(GameCharacter):
    def __init__(self,name):
        super().__init__(name,80)

    def __str__(self):
        return f"[法师] {self.name} | HP: {self.hp}"
        
    def attack(self,target):
        print(f"{self.name}释放了火雷球！")
        target.take_damage(30)



warrior = Warrior("盖伦")
mage = Mage("安妮")

print(warrior)
print(mage)


print("===== 战斗开始 =====")
while warrior.is_alive() and mage.is_alive():
    warrior.attack(mage)
    if mage.is_alive():  # 如果法师还没死，法师再反击
        mage.attack(warrior)
    print("-" * 20)

print("===== 战斗结束 =====")
# 看看最后谁站到了最后
print(warrior)
print(mage)

        
    
    
        

    

    
```

IndentationError: unexpected indent   缩进错误

在 Python 中，当 % 后面跟着多个变量时，必须把它们放在一个元组（用小括号 () 括起来）里

. 核心区别：它们分别给谁看？
__str__ (String)：给用户/玩家看的。
目的：美观、易读。就像游戏里展示给玩家的角色面板。
触发时机：当你使用 print(对象) 或 str(对象) 时。
__repr__ (Representation)：给程序员/开发者看的。
目的：精确、无歧义。它应该尽可能包含重建这个对象所需的所有信息。
触发时机：当你在交互式命令行（如 Jupyter Notebook）里直接敲入对象名回车，或者使用 repr(对象) 时。
. 用你的游戏代码来演示
我们在你刚才写好的 GameCharacter 类里，加上 __repr__ 方法：
python



class GameCharacter:
    def __init__(self, name, hp):
        self.name = name
        self.hp = hp
    
    # 给玩家看的（你已经会了）
    def __str__(self):
        return f"[角色] {self.name} 剩余HP: {self.hp}"
    
    # 给程序员看的（新增）
    def __repr__(self):
        # 理想情况下，repr 返回的字符串，应该能直接复制粘贴去创建对象
        return f"GameCharacter(name='{self.name}', hp={self.hp})"
. 见证奇迹的时刻
假设我们在 Jupyter Notebook 里运行以下代码：
python



warrior = Warrior("盖伦")

# 1. 触发 __str__
print(warrior)  

# 2. 触发 __repr__ (在 Jupyter 里直接敲变量名回车)
warrior 
控制台会输出：
text



[角色] 盖伦 剩余HP: 150           <-- 这是 print(warrior) 触发的 __str__
GameCharacter(name='盖伦', hp=150) <-- 这是直接敲 warrior 触发的 __repr__
💡 一个重要的“保底”机制
如果你只写了 __str__，没写 __repr__，Python 会怎么做？
Python 会非常贴心地用 __str__ 的结果来代替 __repr__。也就是说，直接敲 warrior 也会显示 [角色] 盖伦 剩余HP: 150。
但如果你只写了 __repr__，没写 __str__，Python 会怎么做？
print(warrior) 会显示 GameCharacter(name='盖伦', hp=150)。
最佳实践：在实际开发中，通常两个都写。__str__ 负责漂亮，__repr__ 负责精确。
