#列表

-1）定义：是处理一组有序项目的数据结构
-2）格式：
    列表名 = [元素1，元素2，,,,]
*注意：
     一个列表的数据类型可以各不相同！！！

补充：列表也是可迭代对象！

-3）列表操作：
    1.新增：append9=();extend();insert()
    2.修改：直接通过下标访问进行修改
    3.查询：in；not in；index；count
    4.删除：del；pop；
    5.排序：sort；reverse;
    6.列表推导式：(用来简化上面的操作步骤）
                [表达式 for 变量 in 可迭代对象]
                [表达式 for 变量 in 可迭代对象 if 条件]

```python
li = [1,2,3]
li.append(4)
print(li)
li.extend("five")
print(li)
li.insert(3,6)
print(li)
```

#append整体添加
#extend分散添加
#insert在指定位置添加，若位置被占，则后移一位。

```python
li = [1,2,3]
li[2] = 'a'
print(li)
```

```python
li =[1,2,3]
print(2 in li)
```

#in的返回值是True和False
#index返回的是下标
#count统计出现次数

```python
li = [1,2,3]
del li[2]
print(li)

li.pop()
print(li)

li = [1,2,3,4]
li.pop(2)
print(li)

li = [1,2,3,4]
li.remove(2)
print(li)
```

#pop默认删除最后一个元素，也可以指定下标删除
#remove可以指定元素本身删除
#若存在相同元素，remove默认删除最开始出现的元素

```python
li = [1,3,4,2,5]
li.sort()
print(li)
li.reverse()
print(li)
```

#sort从小到大排序
#reverse默认倒序

```python
[li.append(i) for i in range(1,11) if i % 2==1]
print(li)
```

#元组

-1)基本格式：元组名 = {元素1，元素2，...}
*注意：只有一个元素时，也要加上“ ，”
-2）元组只能查询
-3）应用场景：函数的参数与返回值；格式化输出后面的（）本质上也是一个元组；保护数据

#字典

-1）基本格式：字典名 = （键1：值1，键2：值2，...）
*注意：键具有唯一性，值可以重复，否则前面的会被后面的覆盖
-2）操作：1.查看：变量名[键名](字典没有下标，用键名查询)
         2.修改：通过键名修改
         3.删除：del；clear；pop
         *注意：del会删除字典，clear会保留字典
         4.求长度：len（）
         5.返回键名：keys
         6.返回值：value
         7.返回键值对：item
-3）应用场景：存储信息

#集合

-1）基本格式：集合名 = {元素1，元素2，...}
补充：{}定义的是空字典，set（）定义的才是空集合。
-2）性质：无序性，具体实现方式涉及哈希表；唯一性
-3）操作：1.添加：add（）（添加的是一个整体，一次只能添加v           一个元素！！！）；update（）分散添加，()里是可            迭代对象，如列表，元组。
          2.删除：remove（）选择的是元素删除，会报错；pop           对集合先无序排列，然后删除第一个元素；discard（）不会报错
