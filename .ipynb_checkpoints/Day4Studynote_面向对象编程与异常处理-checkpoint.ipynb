#面向过程
  含义：根据拆解步骤过程来编写代码。

#面向对象
  含义：操作对象解决问题，便于真实的模拟世界
    三大优点：
      -1）更符合人们的思考习惯；
      -2）把复杂的事情简单化；
      -3）把人们从执行者变为指挥者；
    三大特性：
      -1）封装：隐藏对象的属性和实现细节，仅对外公开接口，即将数据和操作结合形成类；
        格式（类）：
          class 类名：
                #属性,名词
                #行为,动词               
      -2）继承：
      -3）多态；
      
#类
  含义：抽象模板。

#对象
  含义：具体事物。
    格式：
      对象名 = 类（）   #创建对象
      对象名.方法名（）  #调用对象

#self关键字
  含义：用于指向对象的实例本身。表示这个类当前对象的引用。
  作用：写在类的函数中，函数通过self来区分是哪个对象调用的它。
    即：def 函数名（self，参数变量）：
    
  补充：若在函数内部调用，则：
    self.方法名（）

#属性
  类外添加和获取对象属性
    1）设置属性：对象名.属性 = 属性值
    2）获取属性：对象名.属性
  类内获取对象属性
    self.属性

#魔法方法
  <1>__init__()
    含义：属于Python内置的函数，在特定的场景下会被自动调用。
    自动调用：当新创建对象时，就会自动触发__init__()魔法方法。
    -1）无参版：类使用后直接初始化属性值，创建对象时自动调用；
    -2）有参版：类使用后先用变量初始化属性，再创建对象时传入属性值进行初始化。

  <2>__str__()
    含义：当用print()函数打印对象时，会自动调用该对象（所在类）的str魔法方法。该方法默认打印的是           对象的地址，所以一般会重写，改为打印对象的谷歌属性值。
    格式：必须写return一个字符串。
  <3>__repr__()


    
#继承
  格式：
    class 子类名（父类名）：
  作用：提高代码复用性，但耦合性增强了。

  <1>单继承
    一个子类只能继承自一个父类。
    基本语法：
      #父类
      class Father（object）：
        pass
      #子类
      class Son（Father）：
        pass

  <2>多继承
    一个类继承多个父类。
    基本语法：
      #父类1
      class Father（object）：
        pass
      #父类2
      class Mother（object）：
        pass
      #子类
      class Son（Father，Mother）：
        pass
    
  继承顺序：从左到右，就近原则。可以使用__mro__属性或类               名.mro()的方法查看调用的先后顺序。
  
  子类重写父类功能：
    含义：当子类的属性或方法与父类相同时，从父类继承下来的  成员可以重新定义！此时会优先调用子类的属性和方法。

  子类调用父类方法：
    若子类仍父想保留父类的行为，则需要在子类中调用父类方法，可以直接使用父类名进行调用。
    即：类名.父类方法名（self）
    也可以：super（）.父类方法名（）    只能访问最近的父类

  多层继承：A继承B，B继承C。

#封装
  将属性和类写到类里。

  私有属性和私有方法
    即不继承给子类。
  格式：
    #私有属性
    __属性名
    #私有方法
    def__方法名（）：
      ...
    使用规则：只能在类的内部使用，若要在外部使用，可以通过公共接口。
    获取私有属性：
      1）获取：  get_xx
      2)修改：  set_xx
    获取私有方法：
      类里再设置一个公有方法来访问私有方法。

#多态
  含义：同一个函数在不同场景下有不同的状态。
  实现条件：
    -1）有继承；
    -2）函数重写
    -3）父类引用指向子类对象


#属性
  <1>对象属性：用__inin__():进行创建，不同的对象有不同的属性值，可以由对象名.属性名进行调用；
  <2>类属性：写在类里，由类下所有对象共有，可以用类名.属性名直接调用。

#文件读写（文本、CSV、JSON）
  #文件的打开
    >普通文件
      仅仅是在一个文件名下的字节流。
    >open()函数
      f = open(filename[,mode[,buffering]])
        -f:open()返回的文件对象
        -filename：文件的字符串名
        -mode：打开模式和文件类型
        -buffering：文件的缓冲区，默认为-1
    >文件的打开模式mode
      1.mode的第一个字母表示对文件的操作：
        “r”表示读
        ‘w’表示写
        ‘x’表示文件不存在的情况下新建文件
        ‘a’表示在文件末尾添加新内容
        ‘+’表示读写模式
      2.第二个字母表示文件类型
        ‘t’是文本类型
        ‘b’是二进制文件
        
  #文件的读写
    >f.write(str)
    >f.read()

  #文件的关闭
    f.close()
