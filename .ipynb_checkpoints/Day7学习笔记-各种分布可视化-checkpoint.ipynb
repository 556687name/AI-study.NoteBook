1.伯努利分布 (Bernoulli Distribution)
它的“人设”：最极致的二元对立。它描述的是单次只有两种可能结果的随机试验（比如：抛一次硬币、一次点击、一次是否生病）。
数学特征：
属于离散型分布。
只有一个参数 𝑝（成功的概率）。
期望是 𝑝；方差是 𝑝(1−𝑝)
在算法与测试中的应用：
二分类问题的基石：逻辑回归（Logistic Regression）的输出本质上就是预测一个伯努利分布的参数 𝑝。
Dropout 正则化：在训练神经网络时，以概率 p 随机“关闭”某些神经元，这就是在采样伯努利分布。
算法测试：用于模拟 A/B 测试中的单次用户转化（点/没点）。

2.高斯分布 / 正态分布 (Gaussian / Normal Distribution)
它的“人设”：自然界的“绝对主宰”。当很多微小的、独立的随机因素叠加在一起时，结果往往呈现高斯分布（中心极限定理）。

在算法与测试中的应用：
高斯朴素贝叶斯：当我们之前讨论的贝叶斯推断遇到连续型特征（如人的身高、体重）时，通常假设这些特征服从高斯分布。
误差分析：在回归算法中，我们通常假设模型的预测误差（残差）服从均值为 0 的高斯分布。
算法测试：用于生成符合真实世界规律的测试数据，比如模拟服务器的响应延迟、用户的年龄分布等。

3.多项分布 (Multinomial Distribution)
它的“人设”：伯努利分布的“超级升级版”。伯努利只有“正/反”两种结果，而多项分布描述的是单次试验有 k 种可能结果的情况。
数学特征：
属于离散型分布。
参数是一个概率向量，且所有概率之和必须为 1。
在算法与测试中的应用：
多分类问题：如果你要做一个模型来识别手写数字（0-9共10个类别），模型输出的就是一个 10 维的概率向量，这本质上就是在预测一个多项分布。
文本分类 / NLP：词袋模型（Bag of Words）统计一篇文章中各个单词出现的次数，这就是典型的多项分布。
算法测试：用于模拟复杂的路由选择、多类别的点击率测试等。

要不要给你一份可直接复制的完整绘图模板，内置配色数组？

```python
import numpy as np 
import matplotlib.pyplot as plt
from scipy.stats import norm

plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False

def plot_bernoulli():
    p = 0.7
    x = np.array([0,1])
    y = np.array([1-p, p])
    plt.bar(x, y, color=['#1f77b4'],alpha=0.7)
    plt.xlabel('Value')
    plt.ylabel('Probability')
    plt.title('Bernoulli Distribution')
    plt.xticks([0, 1])
    plt.show()

def plot_gaussian():
    x = np.linspace(-10, 10, 100)
    y = norm.pdf(x, 0, 2)
    plt.plot(x, y, color='#1f77b4')
    plt.fill_between(x, y, color="#e5360f", alpha=0.3)
    plt.xlabel('Value')
    plt.ylabel('Probability Density')
    plt.title('Gaussian Distribution')
    plt.show()

def plot_multinomial():
    p = np.array([0.1,0.4,0.3,0.2])
    x = np.arange(len(p))
    x_values = np.array(['面1','面2','面3','面4'])
    plt.bar(x, p, color=['#1f77b4'],alpha=0.7)
    plt.xticks(x, x_values)
    plt.xlabel('Value')
    plt.ylabel('Probability')
    plt.title('Multinomial Distribution')
    plt.show()

if __name__ == "__main__":
    plot_bernoulli()
    plot_gaussian()
    plot_multinomial()
```
