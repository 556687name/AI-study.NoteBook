-2）优缺点：编译型语言运行速度快；解释性

#注释
-1）#：单行注释；
-2）三引号：多行注释；
-3）快捷键：Ctrl+/


#print函数

-1）值与值用,隔开

-2）sep=    放在值之后用来确定值与值的分隔符

-3）end用来设定结尾，默认\n换行，可以修改为其它字符串（如" "表示这个print与下一个print的输出用空格隔开

#变量

-1）格式：变量名 = 值

-2）同一个变量可以被反复赋值，且可以是不同的数据类型

#标识符

-1）含义：定义的变量名，函数名

-2）规定：只能用数字，字母，下划线组成。不能用数字开头；不能是关键字；严格区分大小写。

#数值类型

-1）int整型

-2）float浮点型

-3）bool布尔型：True=1，Flase=0.注意严格区分大小写！

-4）complex复数型：固定写法：a + bj

#字符串

-1）特点：要加上引号（单双都行，多行用三引号）

#占位符

-1）作用：生成一定格式的字符串

-2）%s

-3）%d

```python
age = 19
name = 'liuxin'
print("我的名字：%s，我的年龄：%d"%(name,age))
```

-4)%4d:d前加数字表示设置位数，不足的位置为空白，也可以写为%04d把空白变成0，超过的话原样输出。

-5）%f：默认最大小数点后六位小数，遵循四舍五入。

-6）%.4f:注意别忘了小数点，表示设置小数位数。

```python
a = 3.14159
print("%.4f"%a)
```

-7)%%:占位符，print直接输出%，注意print格式

```python
print("我是%%的1%%" %())
```

#f格式化(非常的好用！！！！！）

```python
name = 'liuxin'
age = 19
print(f"我的名字是：{name},我今年{age}了")
```

-4）使用算术运算符时，若有flaot型，则结果也用flaot型表示

#转义字符

-1）\t:制表符，表示空4个字符位置，也称缩进

```python
print("表头\t姓名\t年龄\t成绩")
```

-2)\n:换行符

-3）\r:回车，表示把当前位置移动到本行开头

*当打算把这些转义字符单纯打印出来时，在转义字符前再加上'\’即可。
